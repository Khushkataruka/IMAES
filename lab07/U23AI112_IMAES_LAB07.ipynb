## 1. Propositional Logic
Represent a simple rule such as “If it rains, then the road is wet” using Propositional Logic. Generate its truth table using Python and verify the result.

```python
import itertools
import pandas as pd

print("Propositional Logic: If it rains (P), then the road is wet (Q)")
print("Rule: P -> Q (which is equivalent to ~P | Q)")

# Variables: P (rains), Q (road is wet)
values = [True, False]
truth_table = []

for P, Q in itertools.product(values, values):
    # P -> Q is False only when P is True and Q is False
    implies = not P or Q
    truth_table.append({'P (Rains)': P, 'Q (Road Wet)': Q, 'P -> Q': implies})

df_truth_table = pd.DataFrame(truth_table)
display(df_truth_table)
```

## 2. First-Order Logic
Represent the statements “All humans are mortal” and “Socrates is a human” using First-Order Logic. Write a simple Python program to derive whether Socrates is mortal.

```python
# First-Order Logic deduction using a simple rule-based approach

class FirstOrderLogicSystem:
    def __init__(self):
        self.facts = set()
        self.rules = []

    def add_fact(self, fact):
        self.facts.add(fact)

    def add_rule(self, rule_condition, rule_conclusion):
        # A rule is a function that checks conditions against facts and returns new facts
        self.rules.append((rule_condition, rule_conclusion))

    def deduce(self):
        new_facts = set()
        for condition, conclusion in self.rules:
            # Check for matches in our facts
            entities = {fact.split('(')[1].strip(')') for fact in self.facts if '(' in fact}
            for entity in entities:
                if condition(entity, self.facts):
                    new_facts.add(conclusion(entity))
        
        # Add new facts to knowledge base
        self.facts.update(new_facts)
        return new_facts

fol_system = FirstOrderLogicSystem()

# Fact: Socrates is a human -> human(Socrates)
fol_system.add_fact("human(Socrates)")

# Rule: All humans are mortal -> For all x, human(x) => mortal(x)
def human_condition(entity, facts):
    return f"human({entity})" in facts

def mortal_conclusion(entity):
    return f"mortal({entity})"

fol_system.add_rule(human_condition, mortal_conclusion)

print("Knowledge Base before deduction:", fol_system.facts)
derived = fol_system.deduce()
print("Derived facts:", derived)
print("Knowledge Base after deduction:", fol_system.facts)

if "mortal(Socrates)" in fol_system.facts:
    print("\nConclusion: Socrates is mortal.")
else:
    print("\nConclusion: Cannot derive if Socrates is mortal.")
```

## 3. Probability of Events & 4. Conditional Probability
Consider a small dataset of events such as students who passed/failed and attended/did not attend. Calculate the probability of individual events, joint probability, probability of the union, and conditional probabilities using Python.

```python
# Dataset of students: (Attended, Passed)
# True = Yes, False = No
dataset = [
    (True, True), (True, True), (True, False), (True, True),
    (False, False), (False, True), (False, False), (True, True),
    (True, True), (False, False)
]

total_students = len(dataset)

# Events
# A: Student attended
# B: Student passed
attended = [s for s in dataset if s[0] == True]
passed = [s for s in dataset if s[1] == True]
attended_and_passed = [s for s in dataset if s[0] == True and s[1] == True]
attended_or_passed = [s for s in dataset if s[0] == True or s[1] == True]

# 3. Probability of Events
P_A = len(attended) / total_students
P_B = len(passed) / total_students
P_A_and_B = len(attended_and_passed) / total_students
P_A_or_B = len(attended_or_passed) / total_students

print("--- 3. Probability of Events ---")
print(f"P(A) [Probability of attending] = {P_A}")
print(f"P(B) [Probability of passing] = {P_B}")
print(f"P(A ∩ B) [Joint probability of attending AND passing] = {P_A_and_B}")
print(f"P(A ∪ B) [Union probability of attending OR passing] = {P_A_or_B}")

# Verification of Addition Rule: P(A ∪ B) = P(A) + P(B) - P(A ∩ B)
assert abs(P_A_or_B - (P_A + P_B - P_A_and_B)) < 1e-9

# 4. Conditional Probability
# P(A | B) = P(A ∩ B) / P(B)
# P(B | A) = P(A ∩ B) / P(A)

P_A_given_B = P_A_and_B / P_B if P_B > 0 else 0
P_B_given_A = P_A_and_B / P_A if P_A > 0 else 0

print("\n--- 4. Conditional Probability ---")
print(f"P(A | B) [Probability of attending given they passed] = {P_A_given_B:.3f}")
print(f"P(B | A) [Probability of passing given they attended] = {P_B_given_A:.3f}")
```

## 5. Bayes' Theorem
Implement a simple Bayes' theorem example, such as disease diagnosis. Calculate the probability of a disease after observing a positive test result.

```python
# Bayes' Theorem: P(D | T) = (P(T | D) * P(D)) / P(T)
# D: Has Disease
# T: Positive Test Result

# Given probabilities
P_D = 0.01          # Prior probability of having the disease (1%)
P_T_given_D = 0.99  # Sensitivity: Probability of positive test given disease (99%)
P_T_given_not_D = 0.05 # False positive rate: Probability of positive test given NO disease (5%)

# P(~D)
P_not_D = 1 - P_D

# Total probability of positive test P(T) using Law of Total Probability
# P(T) = P(T | D) * P(D) + P(T | ~D) * P(~D)
P_T = (P_T_given_D * P_D) + (P_T_given_not_D * P_not_D)

# Posterior probability P(D | T)
P_D_given_T = (P_T_given_D * P_D) / P_T

print("--- 5. Bayes' Theorem ---")
print(f"Prior Probability of Disease P(D): {P_D:.4f}")
print(f"Probability of a Positive Test P(T): {P_T:.4f}")
print(f"Posterior Probability of Disease given Positive Test P(D | T): {P_D_given_T:.4f}")
```

## 6. Visualization
Use Matplotlib to create a suitable graph showing the calculated probabilities. Modify one of the input probabilities and observe how the graph changes.

```python
import matplotlib.pyplot as plt
import numpy as np

# Let's observe how the Posterior Probability P(D | T) changes 
# as we vary the Prior Probability P(D) from 0.001 to 0.1 (0.1% to 10%)

prior_probabilities = np.linspace(0.001, 0.1, 100)
posterior_probabilities = []

for prior_p_d in prior_probabilities:
    prior_p_not_d = 1 - prior_p_d
    # Recalculate P(T)
    current_p_t = (P_T_given_D * prior_p_d) + (P_T_given_not_D * prior_p_not_d)
    # Recalculate P(D | T)
    current_p_d_given_t = (P_T_given_D * prior_p_d) / current_p_t
    posterior_probabilities.append(current_p_d_given_t)

plt.figure(figsize=(10, 6))
plt.plot(prior_probabilities, posterior_probabilities, label="P(Disease | Positive Test)", color="blue", linewidth=2)

# Highlight our original P(D) = 0.01 point
original_p_t = (P_T_given_D * 0.01) + (P_T_given_not_D * 0.99)
original_posterior = (P_T_given_D * 0.01) / original_p_t
plt.scatter([0.01], [original_posterior], color="red", zorder=5, label=f"Original P(D)=0.01\nPosterior={original_posterior:.2f}")

# Highlight a modified P(D) = 0.05 point
mod_p_t = (P_T_given_D * 0.05) + (P_T_given_not_D * 0.95)
mod_posterior = (P_T_given_D * 0.05) / mod_p_t
plt.scatter([0.05], [mod_posterior], color="green", zorder=5, label=f"Modified P(D)=0.05\nPosterior={mod_posterior:.2f}")

plt.title("Effect of Prior Disease Probability on Posterior Diagnosis", fontsize=14)
plt.xlabel("Prior Probability of Disease P(D)", fontsize=12)
plt.ylabel("Posterior Probability P(D | Positive Test)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend()
plt.show()
```

### Observation for Exercise 6
The graph shows a strong non-linear relationship between the prior probability of the disease (prevalence in the population) and the confidence in a positive diagnosis (posterior probability). 
If the disease is extremely rare (e.g., `P(D) = 0.01`), a positive test only means a ~16% chance of actually having the disease due to false positives dominating. However, if we modify the prior probability to be higher (e.g., `P(D) = 0.05`), a positive test now means a ~51% chance of having the disease.
